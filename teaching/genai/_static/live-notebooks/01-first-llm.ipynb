## 2. Load the fixed tokenizer and weights

`LocalLM` is a small wrapper in `labs/llm_lab.py`. Its constructor checks the downloaded model identity, loads `AutoTokenizer` and `AutoModelForCausalLM`, sets CPU float32 and evaluation mode, and creates an empty call log. `lm.tokenizer` and `lm.model` expose those two objects.
````python
lm = LocalLM(root / '_models/qwen2.5-1.5b')
print(lm.versions)
print('Model loaded; no question has been generated yet.')
````

## 3. Construct a single request

A list holds messages in order. Each dictionary has a speaker (`role`) and text (`content`). Change only the user text on your second experiment, so you can identify what changed.

````python
messages = [
    {'role': 'system', 'content': 'Explain statistical ideas in one short sentence.'},
    {'role': 'user', 'content': 'What is Monte Carlo error?'},
]
print(json.dumps(messages, indent=2))
````

## 4. Inspect the actual model input

The chat template adds the model-specific role delimiters. `tokenize=False` first returns a string; the next call turns it into integer IDs. `add_special_tokens=False` avoids adding duplicate control tokens. `return_tensors="pt"` asks for PyTorch tensors. Predict: the first shape dimension is one because we have one prompt, not one because we have one token.

````python
from copy import deepcopy
# Freeze the actual input before serialization, not after later turns.
input_snapshot = deepcopy(messages)
text = lm.tokenizer.apply_chat_template(
    input_snapshot, tokenize=False, add_generation_prompt=True)
inputs = lm.tokenizer(text, return_tensors='pt', add_special_tokens=False)
print(text)
print('Token tensor shape:', tuple(inputs.input_ids.shape))
print('First twelve IDs:', inputs.input_ids[0, :12].tolist())
assert inputs.input_ids.ndim == 2 and inputs.input_ids.shape[0] == 1
````

## 5. Expand generation, slicing, and decoding

1. `inference_mode()` suppresses gradient bookkeeping. It does not update weights.
2. `**inputs` passes `input_ids` and `attention_mask` by name to `generate`.
3. `output` contains the original prompt **followed by** generated IDs; its shape is `(1, prompt_length + new_length)`.
4. `output[0, prompt_length:]` selects the one sequence, then removes its prompt. Decoding the whole array would echo the question.
5. `decode` maps the remaining integer IDs to text. `strip` trims surrounding whitespace.
6. `record_generation` saves already computed output, timing and the frozen prompt; it makes **zero additional model calls**. Later labs use `lm(messages)`, which packages these same operations. A token budget limits length, not statistical error or factuality.

````python
import time
started = time.perf_counter()
with lm.torch.inference_mode():
    output = lm.model.generate(
        **inputs, max_new_tokens=64, do_sample=False,
        pad_token_id=lm.tokenizer.eos_token_id)
prompt_length = inputs.input_ids.shape[1]
new_ids = output[0, prompt_length:]
answer = lm.tokenizer.decode(new_ids, skip_special_tokens=True).strip()
logged_answer = lm.record_generation(input_snapshot, text, inputs, output, started, 64)
assert logged_answer == answer
call = lm.calls[-1]
print('Input / complete output / new IDs:', tuple(inputs.input_ids.shape), tuple(output.shape), tuple(new_ids.shape))
print('Raw model answer:', answer)
print('Stop reason:', call['stop_reason'])
assert len(lm.calls) == 1
assert call['prompt_tokens'] == prompt_length
assert 0 < call['generated_tokens'] <= 64
assert call['messages'] == input_snapshot
assert call['serialized_prompt'] == text
````

## 6. Check meaning, then save
